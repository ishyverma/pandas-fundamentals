```python
import numpy as np
import pandas as pd
```

```python
# A panda series is like a column in a table. It is a 1-D array holding data of any type
```

```python
# Series from list
```

```python
country = ['India', 'Australia', 'USA', 'Germany']

pd.Series(country)
```

```python
runs = [31, 24, 56, 78, 100]

runs_series = pd.Series(runs)
```

```python
# Custom index

marks = [67, 57, 89, 100]
subjects = ['Maths', 'English', 'Science', 'Hindi']

pd.Series(marks, index=subjects)
```

```python
shyam_marks = pd.Series(marks, subjects, name='Shyam Marks')
shyam_marks
```

```python
marks = {
    'Maths': 67,
    'English': 57,
    'Science': 89,
    'Hindi': 100
}

marks_series = pd.Series(marks, name='Shyam Marks')
marks_series
```
